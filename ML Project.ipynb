# Visualization

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

### Cancellation VS Check In

```python
cancellation_counts = hotel["is_canceled"].value_counts()
labels = ["Checked-in", "Canceled"]
colors = ["green", "red"]
plt.figure(figsize=(6, 3))
plt.pie(
    cancellation_counts,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=colors,
    wedgeprops={"edgecolor": "black"},
)
plt.title("Booking Cancellation Rate", fontsize=14, fontweight="bold")
```

### Which Season Have The Most Bookings

```python
months_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December",
]
hotel["arrival_date_month"] = pd.Categorical(
    hotel["arrival_date_month"], categories=months_order, ordered=True
)

booking_counts = (
    hotel.groupby(["arrival_date_month", "hotel"], observed=False)
    .size()
    .reset_index(name="count")
)

plt.figure(figsize=(9, 4))
sns.lineplot(
    data=booking_counts,
    x="arrival_date_month",
    y="count",
    hue="hotel",
    marker="o",
    )

plt.title("Seasonality: Total Bookings Count by Month",fontsize=14,fontweight="bold",)
plt.xlabel("Arrival Month", fontsize=12)
plt.ylabel("Number of Bookings", fontsize=12)
plt.xticks(rotation=45)
```

### TOP 10 Countries In Bookings (RESORT VS CITY) Hotels

```python
top_10_countries = hotel["country"].value_counts().head(10).index

df_top10 = hotel[hotel["country"].isin(top_10_countries)]

plt.figure(figsize=(10, 4))
sns.countplot(
    data=df_top10,
    y="country",
    hue="hotel",
    order=top_10_countries,
)

plt.title("Top 10 Origin Countries by Hotel Type ", fontsize=14, fontweight="bold")
plt.xlabel("Number of Bookings", fontsize=12)
plt.ylabel("Country Code", fontsize=12)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Cleaned_Hotel.csv")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
print(df.columns.tolist()) #seeing the available features before using it
```

```python
df['room_mismatch'] = (df['reserved_room_type'] != df['assigned_room_type']).astype(int)
df['deposit_flag'] = (df['deposit_type'] == 2).astype(int)
df['is_online_ta'] = (df['market_segment'] == 'Online TA').astype(int)
df['is_groups'] = (df['market_segment'] == 'Groups').astype(int)
df['is_transient_party'] = (df['customer_type'] == 'Transient-Party').astype(int)
df['is_contract'] = (df['customer_type'] == 'Contract').astype(int)
X = df[['is_repeated_guest', 'previous_cancellations', 'previous_bookings_not_canceled',
        'deposit_flag', 'lead_time', 'room_mismatch', 'is_online_ta', 'is_groups',
        'is_transient_party', 'is_contract', 'adr', 'stays_in_weekend_nights',
        'stays_in_week_nights', 'booking_changes', 'total_of_special_requests',
        'required_car_parking_spaces', 'days_in_waiting_list', 'adults', 'children', 'babies']] #can leave the inside square brackets or can remove it
y = df['is_canceled']

print('Rows used for the model:', len(df))
```

```python
# 80% of the data will be used in training the model while the other 20% will be used to evaluate the model accuracy on unseen data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=19)
#random_state controls the randomness used when train_test_split shuffles and splits your data. Without it, every time you rerun the cell you'd get a different random 80/20 split.
print('Training rows:', len(X_train))
print('Testing rows:', len(X_test))
```

```python

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
print('Coefficient:', model.coef_[0]) #coefficient of X ,only grabs the coefficient for the first feature in your list(is_repeated_guest)
print('Intercept:', model.intercept_) #the intercection point on y-axis
print(f'Equation: probability of cancellation = {model.coef_[0]:.2f} × X + {model.intercept_:.2f}')
# X are all the features
```

```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

results.head(10)
```

```python
print(results['Predicted'].nunique()) #to see the frequency of unqiue predicted values
print(results['Predicted'].value_counts())
```

```python
#evaluation of the model by MSE of linear regression
mse = mean_squared_error(y_test, y_pred)
print('Mean Squared Error:', mse)
```

```python
from sklearn.metrics import accuracy_score

# Predictions on both sets (continuous values, need thresholding)
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

# Threshold at 0.5 to convert into class labels
y_train_pred_class = (y_train_pred >= 0.5).astype(int)
y_test_pred_class = (y_test_pred >= 0.5).astype(int)

# Accuracy on both sets
train_accuracy = accuracy_score(y_train, y_train_pred_class)
test_accuracy = accuracy_score(y_test, y_test_pred_class)

print(f"Train Accuracy: {train_accuracy * 100:.2f}%")
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix

y_pred_class = (y_test_pred >= 0.5).astype(int)

precision = precision_score(y_test, y_pred_class)
recall = recall_score(y_test, y_pred_class)
f1 = f1_score(y_test, y_pred_class)

print(f"Precision: {precision * 100:.2f}%")
print(f"Recall: {recall * 100:.2f}%")
print(f"F1 Score: {f1 * 100:.2f}%")
print(confusion_matrix(y_test, y_pred_class))
```

# Logistic Regression Model

```python
# import SKlearn
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
df['room_mismatch'] = (df['reserved_room_type'] != df['assigned_room_type']).astype(int)
df['deposit_flag'] = (df['deposit_type'] == 2).astype(int)
df['is_online_ta'] = (df['market_segment'] == 'Online TA').astype(int)
df['is_groups'] = (df['market_segment'] == 'Groups').astype(int)
df['is_transient_party'] = (df['customer_type'] == 'Transient-Party').astype(int)
df['is_contract'] = (df['customer_type'] == 'Contract').astype(int)

X = df[['lead_time', 'is_repeated_guest', 'previous_cancellations',
        'previous_bookings_not_canceled', 'deposit_flag', 'adr', 'room_mismatch',
        'stays_in_weekend_nights', 'stays_in_week_nights', 'booking_changes',
        'total_of_special_requests', 'required_car_parking_spaces',
        'days_in_waiting_list', 'adults', 'children', 'babies',
        'is_online_ta', 'is_groups', 'is_transient_party', 'is_contract']]
y = df['is_canceled']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=19)

# --- Scale features (logistic regression needs this for stable or fast convergence) ---
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)   # fit only on train — never let test data leak into scaling
X_test_scaled = scaler.transform(X_test)          # test set only gets transformed, not fit
```

```python
# Create model & train it
classifier = LogisticRegression(max_iter=2000, random_state=19)
classifier.fit(X_train_scaled, y_train)
y_pred = classifier.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score

train_accuracy = accuracy_score(y_train, classifier.predict(X_train_scaled))
test_accuracy = accuracy_score(y_test, classifier.predict(X_test_scaled))

print(f"Train Accuracy: {train_accuracy * 100:.2f}%")
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Precision: {precision * 100:.2f}%")
print(f"Recall: {recall * 100:.2f}%")
print(f"F1 Score: {f1 * 100:.2f}%")
print(confusion_matrix(y_test, y_pred))
```

# KNN Model

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
X = df.drop(["is_canceled", "country"], axis=1)
Y = df["is_canceled"]
X = pd.get_dummies(X)
# df["country"].nunique()
# delete country because it has 177 unique value
```

```python
xTrain, xTest, yTrain, yTest = train_test_split(X, Y, test_size=0.2, random_state=42)
scaler = StandardScaler()
xTrain_scaled = scaler.fit_transform(xTrain)
xTest_scaled = scaler.transform(xTest)
```

```python
Knn_model = KNeighborsClassifier(n_neighbors=11, weights="distance")
Knn_model.fit(xTrain_scaled, yTrain)
y_pred = Knn_model.predict(xTest_scaled)
```

```python
print(f"Accuracy: {accuracy_score(yTest, y_pred) * 100:.2f}%")
print("-" * 20)
print("\nclassification_report\n", classification_report(yTest, y_pred))
print("-" * 20)
print("\nconfusion_matrix\n", confusion_matrix(yTest, y_pred))
```
